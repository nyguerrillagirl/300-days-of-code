# Datacamp - Scikit-Learn

```python
import pandas as pd
```

```python
diabetes_df = pd.read_csv("diabetes.csv")
diabetes_df.head()
```

```python
diabetes_df.columns
```

```python
# How many records in diabetes_df
diabetes_df.shape
```

```python
# remove null BMI entries
mask1 = diabetes_df['BMI'] != 0 
#mask2 = diabetes_df['Glucose'] != 0
diabetes_df = diabetes_df[mask1]
diabetes_df.shape
```

```python
diabetes_bmi0 = diabetes_df[diabetes_df['BMI'].isnull()]
diabetes_bmi0.shape
```

```python
X = diabetes_df.drop("Glucose", axis=1).values
y = diabetes_df["Glucose"].values
print(type(X), type(y))
```

```python
X_bmi = X[:,4] # don't forget we dropped glucose column from diabetes_df
print(y.shape, X_bmi.shape)
```

```python
X_bmi = X_bmi.reshape(-1, 1)
print(X_bmi.shape)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_bmi, y)
plt.ylabel("Blood Glucose (mg/dl)")
plt.xlabel("Body Mass Index")
plt.show()
```

### Fitting a regression model

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_bmi, y)
predictions = reg.predict(X_bmi)
plt.scatter(X_bmi, y)
plt.plot(X_bmi, predictions, color='red')
plt.ylabel("Blood Glucose (mg/dl)")
plt.xlabel("Body Mass Index")
plt.show()
```
