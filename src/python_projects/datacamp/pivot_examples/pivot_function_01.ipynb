```python
import pandas as pd
```

```python
data = { \
	'date': ['2023-07-01', '2023-07-01', '2023-07-02', '2023-07-02'], \
	'city': ['New York', 'Los Angeles', 'New York', 'Los Angeles'], \
	'temperature': [85, 90, 88, 92], \
	'humidity': [70, 65, 72, 60] \
	}
```

```python
df = pd.DataFrame(data)
df
```

```python
pivoted = df.pivot(index='date', columns='city', values='temperature')
pivoted
```

```python
pivoted.loc['2023-07-01', 'Los Angeles']  
```

```python
# Suppose you want both temperature and humidity 
pivoted_multi = df.pivot(index='date', columns='city')
pivoted_multi
```

```python
pivoted_multi2 = df.pivot(index='date', columns='city', values=['temperature', 'humidity'])
pivoted_multi2
```

```python
pivoted_multi['temperature']['New York']  
```

```python
data2 = {
'year': [2023, 2023, 2024, 2024], \
'month': ['Jul', 'Jul', 'Jan', 'Jan'], \
'city': ['New York', 'Los Angeles', 'New York', 'Los Angeles'], \
'temperature': [85, 90, 30, 65] \
}
df2 = pd.DataFrame(data2)
df2
```

```python
pivoted2 = df2.pivot(index=['year', 'month'], columns='city', values='temperature')
pivoted2
```

```python
pivoted3 = df.pivot(index='date', columns='city') 
pivoted3
```

```python
pivoted = df.pivot_table(index='date', columns='city', values='temperature', aggfunc='mean')
pivoted
```
